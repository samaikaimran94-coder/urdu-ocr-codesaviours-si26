```python
import gradio as gr
from transformers import TrOCRProcessor, VisionEncoderDecoderModel
from PIL import Image
import torch

# Load Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Your actual trained model
model_path = "/content/drive/MyDrive/urdu_ocr_data/best_trocr_model"

# Load processor
processor = TrOCRProcessor.from_pretrained(model_path)

# Load trained TrOCR model
model = VisionEncoderDecoderModel.from_pretrained(model_path)

# Evaluation mode
model.eval()

print("✅ Model loaded successfully!")
print("Model path:", model_path)
```

```python
def extract_urdu_text(image):
    """Takes an image, returns extracted Urdu text."""

    if image is None:
        return "Please upload an image"

    pixel_values = processor(
        image,
        return_tensors="pt"
    ).pixel_values

    with torch.no_grad():
        generated_ids = model.generate(pixel_values)

    text = processor.batch_decode(
        generated_ids,
        skip_special_tokens=True
    )[0]

    return text if text else "Could not extract text from this image"
```

```python
interface = gr.Interface(
    fn=extract_urdu_text,

    inputs=gr.Image(
        type="pil",
        label="Upload Urdu Image"
    ),

    outputs=gr.Textbox(
        label="Extracted Urdu Text"
    ),

    title="Urdu OCR — Code Saviours SI-26",

    description=(
        "Upload an image containing Urdu text "
        "and get the extracted text."
    ),

    examples=[]
)

print("✅ Gradio interface created!")
```

```python
interface.launch(share=True)
```
